```python
def f(n,k):
    return 2**(n-k) * ((3**k)/(2**n - 3**k))
```

```python
def fp(n,k):
    return 2**n * ((3/2)**k) /(2**n - 3**k)
```

```python
def check_bound(f, n_max=1000, n_min=1, k_min=0):
    for n in range(n_min, n_max+1):
        for k in range(k_min,n):
            if 2**n -3**k < 0:
                continue
            if not f(n,k) < 2**n:
                print(n,k,'%.3f'%f(n,k),2**n)
                return False
    return True
```

```python
check_bound(f, n_min = 3, k_min=1)
```

```python
check_bound(fp, n_min = 3, k_min=1)
```

```python
def check_bound2(f, n_max=1000, n_min=1, k_min=0):
    for n in range(n_min, n_max+1):
        for k in range(k_min,n):
            if 2**n -3**k < 0:
                continue
            if not f(n,k) < (2**n - 3**k):
                print(n,k,'%.3f'%f(n,k),2**n)
                return False
    return True
```

```python
def check_bound3(f, n_max=1000, n_min=1, k_min=0):
    for n in range(n_min, n_max+1):
        for k in range(k_min,n):
            if 2**n -3**k < 0:
                continue
            if not f(n,k) < 2**(k+k):
                print(n,k,'%.3f'%f(n,k),2**(k+3))
                return False
    return True
```

```python
def f(n,k):
    return 2**(n-k) * ((3**k)/(2**n - 3**k))
```

```python
def ratiof(n,k):
    return f(n,k+1)/f(n,k)
```

```python
N = 500
print(f"n={N}")
print("\t k \t  f(n,k) \t f(n,k+1)/f(n,k)")
previous_ratio = 1000
for n in range(N,N+1):
    max_ratio = 0
    for k in range(1,n):
        if 2**n <= 3**k:
            continue
        if k < n/2:
            continue
        ratio = f(n,k+1)/f(n,k)
        ratio2 = ratiof(n,k+1)/ratiof(n,k)
        ratio3 = (ratiof(n,k+2)/ratiof(n,k+1))/ratio2
        m = ""
        if ratio > previous_ratio:
            m = "*"
        previous_ratio = ratio
        print("\t", k, "\t", "\t", '%.3f'%ratio, ratio2, m)
        max_ratio = max(max_ratio, ratio)
    print(f"n={n}", max_ratio)
```

```python
def print_fa_fb(fa,fb):
    for n in range(1,10):
        for k in range(n):
            print(fa(n,k), fb(n,k))
```

```python
def fa(n,k):
    return 2**n - f(n,k)
def fb(n,k):
    return 2**n + 2**n/(2**n - 3**k) - 3**k * 2**(n-k) / (2**n - 3**k)
```

```python
print_fa_fb(fa,fb)
```
