```python
# 6^k < 2^(n+k) - 6^k; n >= 3; 2^n > 3^k
# n >= ceil(k*log2(3))+1
for n in range(9,200):
    for k in range(n):
        if 2**n <= 3**k:
            continue
        lhs = 3**k + 2**k
        rhs = 2**n
        if not lhs < rhs:
            print("oups", n, k)
```

```python
from math import floor, log2
```

```python
alpha = log2(3)
def t(n):
    return floor(alpha*n)
def s(n):
    return n - t(n)
for n in range(6,200):
    for k in range(n):
        if not 2**n > 3**k:
            continue
        lhs = 2**(s(n))
        rhs = t(n)+2
        if not lhs >= rhs:
            print("oups", n, k)
```
