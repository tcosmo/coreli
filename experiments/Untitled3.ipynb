```python
def fractional_part(x: float) -> float:
    return x - math.floor(x)
```

```python
import math

alpha = math.log2(3/2)
def c_observed(Lmax=200):
    worst = float('inf')
    for ell in range(1, Lmax+1):
        n = int(math.floor(ell/alpha))
        val = abs(ell*math.log(2) - n*math.log(3/2)) * (6**n)
        worst = min(worst, val)
    return worst

print(c_observed())  # should be >= 4/3 ≈ 1.333...
```

```python
def sum(k,l, v = False):
    s = 0
    for j in range(0,k):
        e = 2**(l+j) - 2*3**j
        if v:
            print(e, end=" ")
            if j <= k-2:
                print(e > -0.5*3**j)
            else:
                print(e > -0.5*3**(k-1))
        s += e
    return s
```
